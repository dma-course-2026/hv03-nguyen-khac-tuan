### Bài 5

```python
def in_chuoi_dai_hon(s1, s2):
    if len(s1) > len(s2):
        print(s1)
    elif len(s2) > len(s1):
        print(s2)
    else:
        print(s1)
        print(s2)

in_chuoi_dai_hon("Python", "Data")
in_chuoi_dai_hon("One", "Two")
```

### Bài 27

```python
class Circle:
    def __init__(self, r):
        self.radius = r

    def area(self):
        return 3.14 * self.radius ** 2

c = Circle(5)
print(c.area())
```

### Bài 28

```python
class Hinhchunhat:
    def __init__(self, dai, rong):
        self.dai = dai
        self.rong = rong

    def area(self):
        return self.dai * self.rong

hcn = Hinhchunhat(8, 5)
print(hcn.area())
```

### Bài 29

```python
class Shape:
    def __init__(self):
        pass

    def area(self):
        return 0

class Square(Shape):
    def __init__(self, length):
        Shape.__init__(self)
        self.length = length

    def area(self):
        return self.length * self.length

s = Shape()
sq = Square(4)
print(s.area())
print(sq.area())
```
