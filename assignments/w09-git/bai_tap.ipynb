### Bài 5. Định nghĩa hàm in chuỗi dài hơn (hoặc in cả hai nếu bằng nhau)

```python
def in_chuoi_dai_hon(s1, s2):
    if len(s1) > len(s2):
        print(s1)
    elif len(s2) > len(s1):
        print(s2)
    else:
        print(s1)
        print(s2)

in_chuoi_dai_hon("Python", "Data")
print("--- Trường hợp bằng nhau ---")
in_chuoi_dai_hon("One", "Two")
```

### Bài 9. Hàm tạo dictionary 1..20 và chỉ in các giá trị (values)

```python
def in_values_dict():
    d = {}
    for i in range(1, 21):
        d[i] = i ** 2
    for v in d.values():
        print(v, end=" ")
    print()

in_values_dict()
```

### Bài 10. Hàm tạo dictionary 1..20 và chỉ in các key

```python
def in_keys_dict():
    d = {}
    for i in range(1, 21):
        d[i] = i ** 2
    for k in d.keys():
        print(k, end=" ")
    print()

in_keys_dict()
```

### Bài 27. Class Circle với bán kính và method tính diện tích

```python
import math

class Circle:
    def __init__(self, r):
        self.radius = r

    def area(self):
        return math.pi * (self.radius ** 2)

c = Circle(5)
print(f"Diện tích hình tròn bán kính {c.radius} là: {c.area():.2f}")
```

### Bài 28. Class Hinhchunhat với chiều dài, chiều rộng và method tính diện tích

```python
class Hinhchunhat:
    def __init__(self, dai, rong):
        self.dai = dai
        self.rong = rong

    def area(self):
        return self.dai * self.rong

hcn = Hinhchunhat(8, 5)
print(f"Diện tích hình chữ nhật ({hcn.dai} x {hcn.rong}) là: {hcn.area()}")
```

### Bài 29. Class Shape và class con Square ghi đè method area()

```python
class Shape:
    def __init__(self):
        pass

    def area(self):
        return 0

class Square(Shape):
    def __init__(self, length):
        super().__init__()
        self.length = length

    def area(self):
        return self.length * self.length

s = Shape()
sq = Square(4)
print("Diện tích mặc định của Shape:", s.area())
print("Diện tích của Square cạnh 4:", sq.area())
```
