### Bài 20. Phương trình bậc 2

```python
import math

a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
c = float(input("Nhập hệ số c: "))

if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình có vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print(f"Phương trình có một nghiệm: x = {-c / b}")
else:
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phương trình có nghiệm kép: x1 = x2 = {x}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print(f"Phương trình có 2 nghiệm phân biệt: x1 = {x1}, x2 = {x2}")
```

### Bài 33. Điểm thuộc đường tròn

```python
import math

x0, y0 = map(float, input("Nhập tọa độ tâm x0 y0: ").split())
r = float(input("Nhập bán kính R: "))
xa, ya = map(float, input("Nhập tọa độ điểm A xa ya: ").split())

d = math.sqrt((xa - x0) ** 2 + (ya - y0) ** 2)

if abs(d - r) < 1e-9:
    print("Điểm A nằm trên đường tròn")
elif d < r:
    print("Điểm A nằm trong đường tròn")
else:
    print("Điểm A nằm ngoài đường tròn")
```

### Bài 45. Kiểm tra số nguyên tố

```python
import math

n = int(input("Nhập số n: "))

if n < 2:
    la_nguyen_to = False
else:
    la_nguyen_to = True
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            la_nguyen_to = False
            break

if la_nguyen_to:
    print(f"{n} là số nguyên tố")
else:
    print(f"{n} không phải là số nguyên tố")
```
