### Bài 56. Tính C(k, n) (dùng chương trình con)

```python
def giai_thua(x):
    ket_qua = 1
    for i in range(1, x + 1):
        ket_qua *= i
    return ket_qua

def to_hop(n, k):
    return giai_thua(n) / (giai_thua(k) * giai_thua(n - k))

n = int(input("Nhap n = "))
k = int(input("Nhap k = "))
print(f"C( {n} , {k} ) = {to_hop(n, k)}")
```

### Bài 58. Tìm USCLN và BSCNN của 2 số nguyên dương a và b

```python
def uscln(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def bscnn(a, b):
    return (a * b) // uscln(a, b)

a = int(input("Nhập số nguyên dương a: "))
b = int(input("Nhập số nguyên dương b: "))
print(f"Ước số chung lớn nhất của {a} và {b} là: {uscln(a, b)}")
print(f"Bội số chung nhỏ nhất của {a} và {b} là: {bscnn(a, b)}")
```

### Bài 83. Kiểm tra dãy ngẫu nhiên A có lập thành cấp số cộng không

```python
import random

random.seed(42)
n = int(input("Nhập số lượng phần tử của dãy số, N = "))
A = [random.randint(-100, 100) for _ in range(n)]
print("Mảng A:")
print(*A)

la_csc = True
if n >= 2:
    d = A[1] - A[0]
    for i in range(2, n):
        if A[i] - A[i - 1] != d:
            la_csc = False
            break

if la_csc:
    print("Day la cap so cong")
else:
    print("Day khong la cap so cong")
```
