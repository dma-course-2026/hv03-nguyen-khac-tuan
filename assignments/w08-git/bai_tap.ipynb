### Bài 9

```python
import datetime

ngay_sinh = input("Nhập ngày sinh (y/m/d): ")
nam, thang, ngay = map(int, ngay_sinh.split("/"))
hom_nay = datetime.date.today()
tuoi = hom_nay.year - nam
if (hom_nay.month, hom_nay.day) < (thang, ngay):
    tuoi -= 1
print("Tuổi của bạn là:", tuoi)
```

### Bài 10

```python
so_gio = float(input("Nhập số giờ làm mỗi tuần: "))
luong_gio = float(input("Nhập thù lao mỗi giờ chuẩn: "))

if so_gio <= 44:
    tong_luong = so_gio * luong_gio
else:
    gio_vuot = so_gio - 44
    tong_luong = 44 * luong_gio + gio_vuot * luong_gio * 1.5

print("Tiền thực lĩnh:", int(tong_luong))
```

### Bài 11: Tính Q = sqrt((2 * C * D) / H)

```python
import math

c = 50
h = 30
ds_d = input("Nhập các giá trị D (cách nhau dấu phẩy): ").split(",")
kq = []
for d in ds_d:
    q = round(math.sqrt((2 * c * float(d)) / h))
    kq.append(str(q))
print(",".join(kq))
```

### Bài 22

```python
tong = 0
while True:
    s = input()
    if not s:
        break
    parts = s.split()
    loai = parts[0]
    tien = int(parts[1])
    if loai == "D":
        tong += tien
    elif loai == "W":
        tong -= tien
print(tong)
```

### Bài 23

```python
def kiem_tra_doi_xung(a):
    return a == a[::-1]

def sx_doi_xung(a):
    dem = {}
    for x in a:
        if x in dem:
            dem[x] += 1
        else:
            dem[x] = 1

    so_le = []
    for k in dem:
        if dem[k] % 2 != 0:
            so_le.append(k)

    if len(so_le) > 1:
        return None

    trai = []
    giua = []
    for k in sorted(dem.keys()):
        for i in range(dem[k] // 2):
            trai.append(k)
        if dem[k] % 2 != 0:
            giua.append(k)
    return trai + giua + trai[::-1]

a = list(map(int, input("Nhập dãy số nguyên: ").split()))
if kiem_tra_doi_xung(a):
    print("Dãy đã đối xứng:", a)
else:
    kq = sx_doi_xung(a)
    if kq is not None:
        print("Dãy có khả đối xứng. Sau khi biến đổi:", kq)
    else:
        print("Dãy không có khả đối xứng")
```
