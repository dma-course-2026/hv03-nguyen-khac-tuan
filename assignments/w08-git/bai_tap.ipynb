### Bài 2. Tính giai thừa của một số cho trước

```python
def giai_thua(n):
    if n == 0 or n == 1:
        return 1
    kq = 1
    for i in range(2, n + 1):
        kq *= i
    return kq

n = int(input("Nhập số cần tính giai thừa: "))
print(giai_thua(n))
```

### Bài 9. Tính tuổi dựa trên ngày tháng năm sinh nhập vào (y/m/d)

```python
from datetime import datetime

ngay_sinh_str = input("Nhập ngày sinh (y/m/d): ").strip()
ngay_sinh = datetime.strptime(ngay_sinh_str, "%Y/%m/%d")
hom_nay = datetime.now()
tuoi = hom_nay.year - ngay_sinh.year
if (hom_nay.month, hom_nay.day) < (ngay_sinh.month, ngay_sinh.day):
    tuoi -= 1
print(f"Tuổi hiện tại của bạn là: {tuoi}")
```

### Bài 10. Tính tiền thực lĩnh theo số giờ làm mỗi tuần (chuẩn 44 giờ)

```python
so_gio = float(input("Nhập số giờ làm mỗi tuần: "))
luong_gio = float(input("Nhập thù lao mỗi giờ chuẩn: "))

GIO_CHUAN = 44
if so_gio <= GIO_CHUAN:
    thuc_linh = so_gio * luong_gio
else:
    gio_vuot = so_gio - GIO_CHUAN
    thuc_linh = GIO_CHUAN * luong_gio + gio_vuot * luong_gio * 1.5

print(f"Số tiền thực lĩnh là: {thuc_linh:,.0f}")
```

### Bài 11. Tính giá trị theo công thức Q = √[(2 * C * D) / H]

```python
import math

C = 50
H = 30
chuoi_d = input("Nhập dãy giá trị D (cách nhau bằng dấu phẩy): ")
ket_qua = []
for item in chuoi_d.split(","):
    D = float(item.strip())
    Q = round(math.sqrt((2 * C * D) / H))
    ket_qua.append(str(Q))
print(",".join(ket_qua))
```

### Bài 22. Tính số tiền thực của tài khoản ngân hàng từ nhật ký giao dịch

```python
so_du = 0
print("Nhập nhật ký giao dịch (nhấn Enter dòng trống để kết thúc):")
while True:
    s = input().strip()
    if not s:
        break
    loai, tien = s.split()
    tien = int(tien)
    if loai == "D":
        so_du += tien
    elif loai == "W":
        so_du -= tien
print(so_du)
```

### Bài 23. Kiểm tra dãy số nguyên có khả đối xứng không và biến đổi thành dãy đối xứng

```python
from collections import Counter

def kiem_tra_doi_xung(a):
    return a == a[::-1]

def sx_doi_xung(a):
    dem = Counter(a)
    phan_tu_le = [k for k, v in dem.items() if v % 2 != 0]
    if len(phan_tu_le) > 1:
        return None
    nua_trai = []
    giua = []
    for k in sorted(dem.keys()):
        nua_trai.extend([k] * (dem[k] // 2))
        if dem[k] % 2 != 0:
            giua.append(k)
    return nua_trai + giua + nua_trai[::-1]

a = list(map(int, input("Nhập dãy số nguyên (cách nhau dấu cách): ").split()))
if kiem_tra_doi_xung(a):
    print("Dãy đã là dãy đối xứng:", a)
else:
    ket_qua = sx_doi_xung(a)
    if ket_qua is not None:
        print("Dãy có khả đối xứng. Sau khi biến đổi:", ket_qua)
    else:
        print("Dãy không có khả đối xứng.")
```
